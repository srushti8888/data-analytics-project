```python
import matplotlib.pyplot as plt
import seaborn as sns

missing = df.isnull().sum()

missing = missing[missing > 0]

plt.figure(figsize=(10, 5))

sns.barplot(
    x=missing.values,
    y=missing.index
)

plt.title("Missing Values in Netflix Dataset")
plt.xlabel("Number of Missing Values")
plt.ylabel("Columns")

plt.show()
```
